### Data Loading & Path Resolution
* **Relative Path (`..`)**: The `..` syntax tells Python to look **one folder level up** from where this notebook is saved, allowing us to access the CSV file outside our current directory.
* **Index Column Selection (`index_col=movieId`)**: The CSV contains a pre-existing index as its first column. Setting `index_col=movieId` tells pandas to use that column instead of generating a duplicate, default numeric index.

```python
movies=pd.read_csv('../movies.csv',index_col='movieId') 
movies.head()
```
